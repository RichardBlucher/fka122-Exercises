```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

csv_path = "task6_simulation_values.csv"   # <-- change this
vel_start_col = 2             # first velocity column index (3rd column)


# Load CSV
df = pd.read_csv(csv_path, header=0)  # allows header
data = df.values

# Extract time
times = data[:, 0].astype(float)

# Determine number of particles
ncols = data.shape[1]
vel_cols = ncols - vel_start_col

if vel_cols % 3 != 0:
    raise ValueError("Velocity column count is not divisible by 3. Check vel_start_col.")

N_particles = vel_cols // 3
N_timesteps = data.shape[0]

print(f"Loaded: {N_particles} particles, {N_timesteps} steps.")


vel = np.zeros((N_particles, N_timesteps, 3))

for p in range(N_particles):
    base = vel_start_col + 3*p
    vel[p, :, 0] = data[:, base]
    vel[p, :, 1] = data[:, base + 1]
    vel[p, :, 2] = data[:, base + 2]


dt_array = np.diff(times)
dt = float(np.mean(dt_array))
print("dt =", dt)
```

```python
def vacf_fft_from_vel(vel, dt, chunk_size=None):
    N_particles, N_timesteps, _ = vel.shape
    
    if chunk_size is None:
        chunk_size = N_timesteps
        
    nchunks = N_timesteps // chunk_size
    if nchunks < 1:
        raise ValueError("chunk_size must be <= number of time steps.")
        
    L = chunk_size
    Phi_chunks = np.zeros((nchunks, L))
    
    for cidx in range(nchunks):
        start = cidx * chunk_size
        end = start + chunk_size
        
        C_sum = np.zeros(L)
        
        for i in range(N_particles):
            C_vec = np.zeros(L)
            for comp in range(3):
                a = vel[i, start:end, comp]
                
                # zero-padding to avoid circular wrap-around
                h = np.concatenate([a, np.zeros(L)])
                
                H = np.fft.fft(h)
                S = H * np.conjugate(H)
                s = np.fft.ifft(S).real     # length 2L
                
                denom = np.arange(L, 0, -1)  # [L, L-1, ..., 1]
                C_comp = s[:L] / denom
                C_vec += C_comp
                
            C_sum += C_vec
        
        Phi_chunks[cidx] = C_sum / N_particles

    Phi = np.mean(Phi_chunks, axis=0)
    Phi_std = np.std(Phi_chunks, axis=0) if nchunks > 1 else np.zeros_like(Phi)
    t = np.arange(L) * dt
    
    return t, Phi, Phi_std
```

```python
chunk_size = 2000  

t, Phi, Phi_std = vacf_fft_from_vel(vel, dt, chunk_size=chunk_size)

print("Phi[0] =", Phi[0])
```

```python
Phi /= Phi[0]
plt.figure(figsize=(8,4))
plt.plot(t, Phi, label="VACF Φ(t)/Φ(0)")
#if np.any(Phi_std):
#    plt.fill_between(t, Phi - Phi_std, Phi + Phi_std, alpha=0.25, label="std")
plt.xlabel("time t")
plt.ylabel("Φ(t)/Φ(0)")
plt.title("Velocity Autocorrelation Function (FFT Method)")
plt.xlim(0,2)
plt.grid(True)
plt.legend()
plt.show()
```

```python
D_s = np.trapz(Phi, t) / 3.0
print("D_s (Green-Kubo) =", D_s)
```

```python
# --- Task 7: Spectral function from VACF and D_s extraction ---
# Assumes you already have `t` (1D array of lags) and `Phi` (VACF values at those lags)
# and `dt` (time spacing). If not, run the VACF cell first to get them.

import numpy as np
import matplotlib.pyplot as plt

# If you don't have t, Phi in the notebook namespace, uncomment the demo loader below:
# (Replace with your real data.)
# t, Phi = ...      # from vacf_fft_from_vel(...) or from your previous cell
# dt = t[1] - t[0]

# ---------- Parameters and padding ----------
L = len(Phi)
if 'dt' not in globals():
    dt = t[1] - t[0]
# Per lecture notes: zero-pad to avoid wrap-around. Use Npad = 2*L (recommended).
Npad = 2 * L

# create padded array (pad at the end with zeros)
Phi_padded = np.concatenate([Phi, np.zeros(Npad - L, dtype=float)])

# ---------- Compute discrete cosine-spectrum via rFFT ----------
# The continuous spectral function (as used in Task 7) is:
#   Phi_hat(omega) = 2 * integral_0^infty Phi(t) cos(omega t) dt
# We approximate this with discrete sums. Using rfft:
R = np.fft.rfft(Phi_padded)              # length Npad/2 + 1 complex values
# convert FFT output to spectral function approximation:
# multiply by dt to approximate the time integral, and multiply by 2 for the cosine integral
Phi_hat = 2.0 * dt * np.real(R)         # real-valued spectral function samples

# frequency axes:
freq = np.fft.rfftfreq(Npad, d=dt)      # cycles per time-unit (Hz)
omega = 2.0 * np.pi * freq              # angular frequency ω in rad / time-unit

# ---------- Extract D_s from zero-frequency component ----------
# The relation in the notes: D_s = (1/6) * Phi_hat(omega=0)
Phi_hat_omega0 = Phi_hat[0]             # corresponds to omega = 0
D_from_spectrum = Phi_hat_omega0 / 6.0

# Also compute D from Green-Kubo integral of VACF (numerical integration)
D_from_vacf = np.trapz(Phi, t) / 3.0    # integral_0^T Phi dt / 3

# ---------- Plot spectral function ----------
plt.figure(figsize=(8,4.5))
plt.plot(omega, Phi_hat, label=r'$\hat\Phi(\omega)$')
# zoom hint: often values decay quickly; show small-omega region too
plt.xlim(0, np.max(omega))   # adjust or zoom if you want (e.g., plt.xlim(0,20))
plt.xlabel(r'$\omega$ (rad / time unit)')
plt.ylabel(r'$\hat\Phi(\omega)$')
plt.title('Spectral function (power spectrum) from VACF')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()

# ---------- Optional: plot log-scale spectral tail ----------
plt.figure(figsize=(8,4.2))
plt.semilogy(omega[1:], np.abs(Phi_hat[1:]), label=r'$|\hat\Phi(\omega)|$')
plt.xlabel(r'$\omega$ (rad / time unit)')
plt.ylabel(r'$|\hat\Phi(\omega)|$ (log scale)')
plt.title('Spectral function (log scale) — tails')
plt.grid(True, which='both', ls='--')
plt.tight_layout()
plt.show()

# ---------- Print numeric results and sanity checks ----------
print("N (VACF length) =", L, ", zero-padded to Npad =", Npad)
print(f"Phi_hat(omega=0) = {Phi_hat_omega0:.6g}")
print(f"D_s (from spectrum, 1/6 * Phi_hat(0)) = {D_from_spectrum:.6g}")
print(f"D_s (from VACF integral, (1/3) * ∫ Phi dt ) = {D_from_vacf:.6g}")
print(f"Relative difference = {(D_from_spectrum - D_from_vacf) / (abs(D_from_vacf) + 1e-30):.3g}")

# ---------- (Optional) Save arrays to file for plotting/report ----------
# np.savetxt("Phi_hat_vs_omega.txt", np.column_stack([omega, Phi_hat]),
#            header="omega  Phi_hat")
# np.savetxt("Phi_vs_t.txt", np.column_stack([t, Phi]), header="t  Phi(t)")
```
