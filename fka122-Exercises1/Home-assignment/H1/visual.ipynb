```python

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

data = pd.read_csv('energies-task1.csv', header=None)
data_array = data.to_numpy()
#print(data_array)
N = data_array.shape
unitcellvolume = np.pow(data_array[:,0],3)
energypercell = data_array[:,1] / 64
fit = np.polyfit(unitcellvolume, energypercell, 2)
print(fit)
x = np.linspace(unitcellvolume[0], unitcellvolume[-1], 1000)
y= np.polyval(fit, x)
print(x[np.argmin(y)])
print(N)
plt.scatter(unitcellvolume,energypercell , label='Potential Energy')
plt.plot(x,y, label='Fitted Curve')
plt.xlabel('Unit Cell Volume (Å³)')
plt.ylabel('Energy per Unit Cell (eV)')
plt.title('Energy per Unit Cell vs. Unit Cell Volume')
plt.legend()
plt.grid()
plt.show()
```
