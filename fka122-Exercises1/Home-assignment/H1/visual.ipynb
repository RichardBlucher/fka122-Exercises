```python

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

data = pd.read_csv('energies-task1.csv', header=None)
data_array = data.to_numpy()
#print(data_array)
N = data_array.shape
unitcellvolume = np.pow(data_array[:,0],3)
energypercell = data_array[:,1] / 64
fit = np.polyfit(unitcellvolume, energypercell, 2)
print(fit)
x = np.linspace(unitcellvolume[0], unitcellvolume[-1], 1000)
y= np.polyval(fit, x)
print(x[np.argmin(y)])
print(N)
plt.scatter(unitcellvolume,energypercell , label='Potential Energy')
plt.plot(x,y, label='Fitted Curve')
plt.xlabel('Unit Cell Volume (Å³)')
plt.ylabel('Energy per Unit Cell (eV)')
plt.title('Energy per Unit Cell vs. Unit Cell Volume')
plt.legend()
plt.grid()
plt.show()
```

```python
data = pd.read_csv('task5a_simulation_values.csv', header=None)
data_array = data.to_numpy()

C_l=[]
C_l_temp=0

for i in range(len(data_array[:,0])):
    C_l_temp=0
    for j in range (len(data_array[:,0])-i-1):
        
        C_l_temp += (data_array[j,1]*data_array[j+i,1])
    C_l_temp /= (len(data_array[:,0])-i)
    C_l.append(C_l_temp)
    C_l_temp=0
        



plt.plot(data_array[:,0], C_l, label='Velocity Autocorrelation Function')
plt.xlabel('Time Lag (ps)')
plt.ylabel('C_l')
plt.title('Velocity Autocorrelation Function over Time Lag')
plt.legend()
plt.grid()
plt.show()
plt.plot(data_array[:,0], data_array[:,1], label='MSD over Time')
plt.xlabel('Time (ps)')
plt.ylabel('Mean Squared Displacement (Å²)')
plt.title('Mean Squared Displacement over Time')
plt.legend()
plt.grid()
plt.show()
```
