```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data=pd.read_csv('block_averages.csv', header=None)
data=np.array(data)

plt.plot(data)
plt.xlabel('Block Size (x100)')
plt.ylabel('Block Average')
plt.title('Block Averages vs Block Size')
plt.grid()
plt.xlim(0, 400)
plt.show()
```
