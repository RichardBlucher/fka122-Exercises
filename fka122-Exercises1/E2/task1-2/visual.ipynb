```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

wo_importance = pd.read_csv('results_wo_importance.csv', header=None).to_numpy()




fig, ax = plt.subplots()
ax.errorbar(wo_importance[:,0], wo_importance[:,1], yerr=wo_importance[:,2], label='wo', fmt='o')
ax.set_xscale('log')
plt.title('Estimated integral and errors for MC without importance sampling')
plt.xlabel('Number of samples')
plt.ylabel('Estimated value of integral')
plt.show()
```

```python
w_importance = pd.read_csv('results_w_importance.csv', header=None).to_numpy()




fig, ax = plt.subplots()
ax.errorbar(w_importance[:,0], w_importance[:,1], yerr=w_importance[:,2], label='w', fmt='o')
ax.set_xscale('log')
plt.title('Estimated integral and errors for MC with importance sampling')
plt.xlabel('Number of samples')
plt.ylabel('Estimated value of integral')
plt.show()
```
